```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
# Convert to DataFrame
df = pd.DataFrame(cluster_mode_data)

# Create a new column for hue, combining keys_million and value_size_bytes
df['data_config'] = df.apply(lambda row: f"{row['keys_million']:.2f}M Keys, {row['value_size_bytes']} Bytes/Value", axis=1)

# Set Seaborn style for a professional look
sns.set_theme(style="whitegrid", palette="viridis") # Changed palette for more distinct colors

# Create the plot
plt.figure(figsize=(12, 7)) # Increased figure size for better clarity with multiple lines
line_plot = sns.lineplot(
    data=df,
    x="threads",
    y="bg_save_time",
    hue="data_config", # Use the new combined column for different lines
    marker='o',        # Add markers for data points
    linewidth=2.5,     # Thicker line for emphasis
    dashes=False       # Ensure solid lines
)

# Add titles and labels
plt.title('Valkey Cluster BGSAVE Time vs. Number of Threads by Data Configuration', fontsize=16)
plt.xlabel('Number of Threads (rdb-snapshot-threads)', fontsize=12)
plt.ylabel('BGSAVE Time (seconds)', fontsize=12)

# Customize ticks and grid
# Ensure all thread numbers are shown as ticks, but only for the actual range present in data
plt.xticks(sorted(df['threads'].unique()), fontsize=10)
plt.yticks(fontsize=10)
plt.grid(True, linestyle='--', alpha=0.7) # Enhance grid visibility

# Add data labels on top of points for clarity for each line
# Iterate through each unique data_config to place labels accurately
for config_name in df['data_config'].unique():
    subset = df[df['data_config'] == config_name]
    for x, y in zip(subset['threads'], subset['bg_save_time']):
        plt.text(x, y + 0.1, f'{y:.3f}', ha='center', va='bottom', fontsize=9, color='dimgray') # Use dimgray for text color for better contrast

# Adjust y-axis limits to start a bit below the min value and end a bit above the max
# Calculate global min/max for all data points
min_y = df['bg_save_time'].min()
max_y = df['bg_save_time'].max()
plt.ylim(min_y * 0.9, max_y * 1.1)

# Place the legend outside the plot to avoid overlapping with data
plt.legend(title='Data Configuration', bbox_to_anchor=(1.02, 1), loc='upper left', borderaxespad=0.)

# Ensure layout is tight to prevent labels/legend from being cut off
plt.tight_layout()

# Save the plot
plot_filename = "bg_save_time_vs_threads_by_config.png"
plt.savefig(plot_filename, bbox_inches='tight')

print(f"Plot saved as {plot_filename}")
```
